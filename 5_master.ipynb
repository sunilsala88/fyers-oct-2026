```python
import pandas as pd


MASTER_URL = 'https://public.fyers.in/sym_details/NSE_CM.csv'

# The Fyers master file has no header row, so name the 21 columns ourselves
COLUMNS = [
    'fytoken', 'symbol_details', 'instrument_type', 'lot_size', 'tick_size', 'isin',
    'trading_session', 'last_update_date', 'expiry_date', 'symbol', 'exchange', 'segment',
    'scrip_code', 'underlying_symbol', 'underlying_scrip_code', 'strike_price', 'option_type',
    'underlying_fytoken', 'reserved_1', 'reserved_2', 'reserved_3',
]


def get_master(url=MASTER_URL, save_path='NSE_CM.csv'):
    df = pd.read_csv(url, header=None, names=COLUMNS, dtype={'fytoken': str, 'underlying_fytoken': str})
    df.to_csv(save_path, index=False)
    print(f'Saved {len(df)} symbols to {save_path}')
    return df


master_df = get_master()
master_df.head()
```
